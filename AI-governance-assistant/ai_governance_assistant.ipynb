```python
from openai import OpenAI
client = OpenAI()

prompt = """
You are an AI governance expert in healthcare.

Given the following AI system, generate a structured risk assessment checklist.

AI System:
An AI model that predicts hospital readmissions using patient data.

Provide:
- Data risks
- Model risks
- Clinical risks
- Ethical risks
- Regulatory risks
"""
```

```python
response = client.chat.completions.create(
    model="gpt-4.1-mini",
    messages=[
        {"role": "user", "content": prompt}
    ]
)

print(response.choices[0].message.content)
```

```python
def generate_governance_checklist(system_description):
    prompt = f"""
    You are an AI governance expert in healthcare.

    Analyze this AI system:
    {system_description}

    Provide a structured checklist including:
    - Data risks
    - Model risks
    - Clinical risks
    - Ethical risks
    - Regulatory risks
    """

    response = client.chat.completions.create(
        model="gpt-4.1-mini",
        messages=[{"role": "user", "content": prompt}]
    )

    return response.choices[0].message.content
```

```python
result = generate_governance_checklist(
    "AI tool for detecting cancer from medical images"
)

print(result)
```

This project demonstrates how generative AI can support AI governance in healthcare by generating structured risk assessment checklists for AI systems.
